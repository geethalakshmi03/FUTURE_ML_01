```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("SuperStore_Sales_Dataset.csv")
```

```python
df = df.drop(columns=["ind1", "ind2"])
df["Returns"] = df["Returns"].fillna("No Return")
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
```

```python
daily_sales = df.groupby("Order Date")["Sales"].sum().reset_index()
```

```python
daily_sales["Day_Number"] = range(len(daily_sales))
```

```python
X = daily_sales[["Day_Number"]]
y = daily_sales["Sales"]

model = LinearRegression()
model.fit(X, y)
```

```python
daily_sales["Predicted_Sales"] = model.predict(X)
```

```python
mae = mean_absolute_error(y, daily_sales["Predicted_Sales"])
rmse = np.sqrt(mean_squared_error(y, daily_sales["Predicted_Sales"]))
r2 = r2_score(y, daily_sales["Predicted_Sales"])

print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
plt.figure(figsize=(12,5))
plt.plot(daily_sales["Order Date"], daily_sales["Sales"], label="Actual Sales")
plt.plot(daily_sales["Order Date"], daily_sales["Predicted_Sales"], label="Predicted Sales")
plt.title("Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```
