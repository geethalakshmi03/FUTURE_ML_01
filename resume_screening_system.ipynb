```python
import pandas as pd

data = {
    "Candidate": [
        "Rahul",
        "Priya",
        "Arjun",
        "Sneha",
        "Kiran"
    ],
    "Resume Skills": [
        "Python, Machine Learning, SQL",
        "Java, Spring Boot, SQL",
        "Python, Data Analysis, Excel",
        "Machine Learning, Deep Learning, Python",
        "HTML, CSS, JavaScript"
    ]
}

df = pd.DataFrame(data)
df
```

```python
job_skills = ["Python", "Machine Learning", "SQL"]

def calculate_score(resume):
    skills = [s.strip() for s in resume.split(",")]
    matched = [skill for skill in job_skills if skill in skills]
    missing = [skill for skill in job_skills if skill not in skills]
    return len(matched), ", ".join(missing)

df[["Score", "Missing Skills"]] = df["Resume Skills"].apply(
    lambda x: pd.Series(calculate_score(x))
)

df = df.sort_values(by="Score", ascending=False)

print(df)
```

```python
print("Resume Screening System Completed Successfully!")
print("\nTop Candidate:", df.iloc[0]["Candidate"])
print("Top Score:", df.iloc[0]["Score"])
```
